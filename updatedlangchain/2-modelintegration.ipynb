### MODELS INTEGRATION WITH OPENAI, GOOGLE GEMINI, AND GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser

model=init_chat_model("google_genai:gemini-3.5-flash-lite")

chain = model | StrOutputParser()
response=chain.invoke("Why do parrots talk?")
print(response)
```

### Google Model Integration

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

parser = model | StrOutputParser()

print(parser.invoke("Does openAI give free API keys like google do?"))
```

### Groq Model Integration - Method 1

```python
import os
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model(model="groq:openai/gpt-oss-120b")

response = model | StrOutputParser()

print(response.invoke("How does groq compare to Gemini and OpenAI models?"))
```

### Groq Integration - Method 2

```python
import os
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model = ChatGroq(model="openai/gpt-oss-120b")

chain = model | StrOutputParser()
otpt = chain.invoke("Wassup cuh")
print(otpt)
```

### Batch

Batching is a collection of independent requests to a model, which can significantly improve perfomance and reduce costs, as the processing can be done in parallel.

```python
responses = chain.batch([
    "Did human beings evolve to human beings from one single organism or was it like multiple species evolving into human simulatenously across the globe",
    "Every legend or culture has some sort of creatures like dragons or other leviathan class life forms, so does that mean they did actually exist? or is it a case of mass hysteria?",
    "Why are scientists still trying tor revive the already extinct species, even dinosaurs? What problem is it going to solve?"
])
for response in responses:
    print(response)
```
